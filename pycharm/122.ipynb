```python
import time

def simple_progress(total=100):
    """简单进度条"""
    for i in range(total + 1):
        percent = i / total * 100
        bar ='\033[31m' +'█' * i + '\033[0m' + '\033[37m'+'░' * (total - i)+'\033[0m'
        print(f"\r进度: {bar}{'\033[31m'}{percent:.1f}%{'\033[0m'}", end='')
        time.sleep(0.05)
    print()  # 换行

# 使用
simple_progress(50)
```
